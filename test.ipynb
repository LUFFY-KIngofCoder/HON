```python
import pandas as pd

df = pd.DataFrame({
    'category': ['A', 'A', 'B', 'B'],
    'sales': [100, 150, 200, 50],
    'quantity': [2, 3, 5, 1]
})

# Apply multiple functions to specific columns
df[['sales', 'quantity']].agg(['sum', 'mean', 'max'])
```

```python
df.groupby('category').agg({
    'sales': ['sum', 'mean'],
    'quantity': 'max'
})
```

```python
df.groupby('category')['sales'].agg(
    total_sales='sum',
    avg_sales= 'mean'
)
```
